```python
from fileformer import FileFormer, ByteLevelTokenizer
import torch
```

```python
ff = FileFormer(257, 256, 4, 4, 0.0)
ff.load_state_dict(
    torch.load("/Users/daniilogorodnikov/PycharmProjects/Notus/model_flash_epoch_9.pt", map_location="cpu")[
        'model_state_dict'])
ff.eval()
```

```python
tokenizer = ByteLevelTokenizer()
```

```python
from fileformer import Engine
```

```python
eng = Engine(42, ff, tokenizer)
```

```python
with open('enwik8', 'rb') as f:
    data_f = f.read(1024).hex()
```

```python
out = eng.compress(data_f)
```
