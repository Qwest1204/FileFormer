```python
import torch
```

```python
checkpoint = torch.load('/Users/daniilogorodnikov/model-epoch=02-step=110000.ckpt', map_location=torch.device('cpu'))
```

```python
# Извлечение state_dict модели
if 'state_dict' in checkpoint:
    model_state = checkpoint['state_dict']
else:
    raise KeyError("Ключ 'state_dict' не найден в checkpoint — проверьте формат файла.")
```

```python
# Вывод структуры (сортируем ключи для удобства)
print("Структура модели из Lightning Checkpoint (state_dict):")
for name, param in sorted(model_state.items()):
    print(f"{name}: {param.shape}")
```
